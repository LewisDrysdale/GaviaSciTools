```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np
import matplotlib as mpl

names=['SD041-KANGGLAC-3-MIIPUGUT',
       'SD041-KANGGLAC-EAST OF UUNARTAAJIIT',
       'SD041-KANG-GLAC-GEO-KVIOQ',
       'SD041-KANG-GLAC-GEO-RYBERG',
       'SD041-KANGGLAC-INNER SHELF NE OF KANGERUSSUAQ TROUGH',
       'SD041-KANG-GLAC-GEO-NMLSFJORD',
       'SD041-KANGGLAC-PAATTUULAAJIT IKERTIVAAT',
       'SD041-KANGGLAC-ROSENBORG-GLETSJER',
       'SD041-KANGGLAC-KVIOQ']
for name in names:

    indir=os.path.join('E:\\Kang-Glac\\data\\Gavia_AUV\\data',name,'files\\processed-log-data')
    print("working on " + name)
    
    
    df=pd.read_csv(os.path.join(indir,'auvdata.csv'))
    
    import pandas as pd
    import numpy as np
    
    # Define the columns to keep
    columns_to_keep = [
        'timestamp',
        'time-nav',
        'altitude',
        'speed',
        'Heading',
        'Depth',
        'Lat',
        'Lon',
        'time-ctd',
        'conductivity',
        'pressure',
        'salinity',
        'sound-velocity',
        'temperature_x',
        'o2-concentration',
        'raw-temp',
        'temperature_y',
        'time-eco',
        'chl-raw',
        'chl-ref',
        'ntu-raw',
        'ntu-ref',
        'thermistor',
        'chl-corrected',
        'ntu-corrected'
    ]
    
    # Filter the dataframe to keep only the desired columns
    df_filtered = df[columns_to_keep].copy()
    
    # Drop rows with missing conductivity
    df_filtered = df_filtered[df_filtered['conductivity'].notna()]
    
    # Convert DDM (e.g., 2915.3155) to Decimal Degrees
    def ddm_to_dd(ddm):
        try:
            ddm = float(ddm)
            degrees = int(ddm // 100)
            minutes = ddm % 100
            return degrees + minutes / 60
        except:
            return np.nan
    
    # Apply conversion
    df_filtered['Lat'] = df_filtered['Lat'].apply(ddm_to_dd)
    df_filtered['Lon'] = df_filtered['Lon'].apply(ddm_to_dd)
    # Optional: drop rows with invalid (NaN) Lat/Lon after conversion
    df_filtered = df_filtered.dropna(subset=['Lat', 'Lon'])
    
    # Final cleaned DataFrame is df_filtered
    df_filtered.to_csv(os.path.join(r'C:\Users\sa01ld\OneDrive - SAMS\Projects\Kang-Glac\data_for_Laura',name+'_filtered_data.csv'),index=False)
                   
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Prepare data: remove rows with missing lat/lon/depth/temperature
plot_df = df_filtered.dropna(subset=['Lat', 'Lon', 'Depth', 'temperature_x'])

# Extract variables
x = plot_df['Lon']
y = plot_df['Lat']
z = plot_df['Depth']  # Keep as-is
c = plot_df['temperature_x']

# Create 3D plot
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot
sc = ax.scatter(x, y, z, c=c,
                cmap='viridis',
                s=30,
                alpha=0.8,
                vmin=-1,vmax=0)   

# Invert Z axis (so depth increases downward)
ax.invert_zaxis()

# Colorbar for temperature
cbar = fig.colorbar(sc, ax=ax, shrink=0.5, aspect=10)
cbar.set_label('Temperature (°C)', fontsize=12)

# Labels and title
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_zlabel('Depth (m)')
ax.set_title('3D Scatter Plot: Temperature Kivioq')

plt.tight_layout()
plt.show()
```
