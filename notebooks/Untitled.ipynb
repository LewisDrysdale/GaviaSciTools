```python
import pandas as pd
import csv
import matplotlib.pyplot as plt
import os

inpath=r'E:\GAVIA\Projects\Morgan_PhD\TM_PHD_Spring2025\USBL_log_files'
csv_file=os.path.join(inpath,'ExportData_2025May01_120805_FF_DAY3.csv')
target_source = 'LatLongObs'

# Step 1: Read the first 3 headers
with open(csv_file, 'r') as f:
    header_lines = [next(f).strip() for _ in range(4)]
    all_headers = [next(csv.reader([line])) for line in header_lines]

# Map source name to the correct header line
# You can hard-code or detect it; here we hard-code:
source_header_map = {
    'LatLongObs': all_headers[3],  # assuming LatLongObs uses header line 2 (index 1)
    # Add other source-name mappings if needed
}

selected_header = source_header_map[target_source]

# Step 2: Read and filter rows for that source
valid_rows = []
with open(csv_file, 'r') as f:
    for i, line in enumerate(f):
        if i < 3:
            continue  # Skip the header lines

        row = next(csv.reader([line.strip()]))

        if not row or row[0] != target_source:
            continue  # Only keep rows starting with the desired source name

        if len(row) != len(selected_header):
            continue  # Skip malformed rows

        valid_rows.append(dict(zip(selected_header, row)))

# Step 3: Create DataFrame
df = pd.DataFrame(valid_rows)
df.tail(10)

df = df.drop(columns=['AssocUID', 'ParentUID', 'FilterUID', 'ObsUID', 'ParentName', 'SourceUID', 'LatLongObs'])
# Now filter by a string match in a specific column
filtered_df = df[df['SourceName'] == '2207']
filtered_df.head(100)
```

```python
from mpl_toolkits.mplot3d import Axes3D  # Needed for 3D plotting

df_scatter = filtered_df[:10]

# Assume df is your cleaned DataFrame from LatLongObs
# Replace with your actual column names
x_col = 'Longitude'  # or 'Longitude'
y_col = 'Latitude'  # or 'Latitude'
z_col = 'Altitude'  # or 'Altitude'

# Create a 3D scatter plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(df_scatter['Longitude'], df_scatter['Latitude'], df_scatter['Altitude'], c='blue', marker='o')

# Axis labels
ax.set_xlabel(x_col)
ax.set_ylabel(y_col)
ax.set_zlabel(z_col)
ax.set_title(f'3D Scatter Plot: {x_col}, {y_col}, {z_col}')

plt.tight_layout()
plt.show()
```
